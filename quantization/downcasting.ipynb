```python
import torch

torch.__version__
```

```python
f32_tensor = torch.rand(1000)
f32_tensor,  f32_tensor.dtype
```

```python
torch.dot(f32_tensor,  f32_tensor)
```

```python
# Down casting

bf16_tensor = f32_tensor.to(dtype=torch.bfloat16)

bf16_tensor ,  bf16_tensor.dtype

torch.dot(bf16_tensor, bf16_tensor)


```

```python
import tensorflow as tf

f32_tensor = tf.random.uniform(shape=(5,))
f32_tensor
```

```python
bf16_tensor = tf.cast(f32_tensor, tf.bfloat16)

bf16_tensor
```

```python
tf.tensordot(bf16_tensor, bf16_tensor, axes=1)
```
